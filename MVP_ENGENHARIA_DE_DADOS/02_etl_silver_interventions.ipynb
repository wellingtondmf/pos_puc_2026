```python
interventions = (clinical_parsed
    .select("nct_id",F.explode_outer(p["armsInterventionsModule"]["interventions"]).alias("intervention"))
    .select(F.upper(F.trim("nct_id")).alias("nct_id"),F.upper(F.trim("intervention.type")).alias("intervention_type"),
            F.regexp_replace(F.trim("intervention.name"),r"\s+"," ").alias("intervention_name"),
            F.trim("intervention.description").alias("description"))
    .filter(F.col("intervention_name").isNotNull()).dropDuplicates())
```

```python
interventions.write.format("delta")\
                   .mode("overwrite")\
                   .option("overwriteSchema", "true")\
                   .saveAsTable(f"{CATALOG}.{SCHEMA}.slv_interventions")
```
