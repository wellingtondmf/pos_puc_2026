```python
conditions = (clinical_parsed
    .select("nct_id",F.explode_outer(p["conditionsModule"]["conditions"]).alias("condition_name"))
    .select(F.upper(F.trim("nct_id")).alias("nct_id"),F.regexp_replace(F.trim("condition_name"),r"\s+"," ").alias("condition_name"))
    .filter(F.col("condition_name").isNotNull())
    .dropDuplicates())
```

```python
conditions.write.format("delta")\
                .mode("overwrite")\
                .option("overwriteSchema", "true")\
                .saveAsTable(f"{CATALOG}.{SCHEMA}.slv_conditions")
```
